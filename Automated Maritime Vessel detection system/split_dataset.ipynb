```python
import os
import random
import shutil

# Define paths
raw_dir = "data/raw"
labels_dir = "data/labels"
base_out = "data/processed"

# Create the standard ML target structure
for split in ["train", "val"]:
    os.makedirs(f"{base_out}/images/{split}", exist_ok=True)
    os.makedirs(f"{base_out}/labels/{split}", exist_ok=True)

# Get and shuffle all matching image names
images = [f for f in os.listdir(raw_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
random.seed(42)
random.shuffle(images)

# Split 80% for training, 20% for validation
split_idx = int(len(images) * 0.8)
train_images = images[:split_idx]
val_images = images[split_idx:]

def move_files(file_list, split_name):
    for img_name in file_list:
        base_name = os.path.splitext(img_name)[0]
        xml_name = f"{base_name}.xml"
        
        # Paths
        src_img = os.path.join(raw_dir, img_name)
        src_xml = os.path.join(labels_dir, xml_name)
        
        # Move image
        shutil.copy(src_img, f"{base_out}/images/{split_name}/{img_name}")
        
        # Move XML label if it exists
        if os.path.exists(src_xml):
            shutil.copy(src_xml, f"{base_out}/labels/{split_name}/{xml_name}")

# Execute the splits
move_files(train_images, "train")
move_files(val_images, "val")

print(f"Dataset successfully split! Trained items: {len(train_images)}, Validation items: {len(val_images)}")
```
