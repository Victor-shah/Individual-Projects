```python
import pandas as pd
import random

# Load the base encounter file we generated
df_encounters = pd.read_csv("fact_encounters.csv")

# --- 1. EXTRACT DEMOGRAPHICS TABLE (Dim_Patients) ---
# Isolate unique Patient IDs and assign realistic regional demographic details
unique_patients = df_encounters["Patient_ID"].unique()

demographics_data = []
nationalities = ['UAE National', 'Expats (Indian)', 'Expats (UK)', 'Expats (Philippine)', 'Expats (Arab)']

for pat_id in unique_patients:
    demographics_data.append({
        "Patient_ID": pat_id,
        "Age": random.randint(1, 85),
        "Gender": random.choice(["Male", "Female"]),
        "Nationality_Group": random.choice(nationalities),
        "Primary_Clinic_Location": "Dubai Festival City"  # Aligning to corporate hub
    })

df_demographics = pd.DataFrame(demographics_data)

# --- 2. EXTRACT CLINICAL DIAGNOSTICS TABLE (Dim_Diagnostics) ---
# Standard clinical dimension mapping ICD-10 keys to SNOMED codes
df_diagnostics = df_encounters[["ICD10_Code", "SNOMED_Description"]].drop_duplicates().reset_index(drop=True)


# --- 3. CLEAN UP FACT ENCOUNTERS (To maintain strict Star Schema mapping) ---
# Remove the text description from the main encounter fact table to keep it highly optimized. 
# The database will link to Dim_Diagnostics via the ICD10_Code foreign key.
df_encounters_cleaned = df_encounters.drop(columns=["SNOMED_Description"])


# --- SAVE CODES ---
df_demographics.to_csv("dim_demographics.csv", index=False)
df_diagnostics.to_csv("dim_diagnostics.csv", index=False)
df_encounters_cleaned.to_csv("fact_encounters.csv", index=False) # Overwrites with optimized clean fact table

print("Normalization complete!")
print(f"Extracted {len(df_demographics)} unique patient profiles into 'dim_demographics.csv'")
print(f"Extracted unique industry diagnostic standards into 'dim_diagnostics.csv'")
```
